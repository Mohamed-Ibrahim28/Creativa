```python
degree=int(input("please enter your degree : "))
while True:
    if degree<0 or degree>100:
        print("please enter valid degree")
        degree=int(input("please enter your degree : "))
    elif degree in range(90 ,101):
            print("A")
            break
    elif degree in range(80,90):
            print("B")
            break
    elif degree in range(70,80):
            print("C")
            break
    elif degree in range(60,70):
            print("D")
            break
    elif degree<60:
            print("F")
            break
```
