```python
import os
import random 
def create_folder(folder_path,num_files):
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
    for i in range(num_files):
        file_name = os.path.join(folder_path,f"file_{i+1}")
        if not os.path.exists(file_name):
            os.makedirs(file_name)
    files=os.listdir(folder_path)
    print(f"number of files in folder is {len(files)}")
forlder_path="F:\\Courses\\Creativa\\session one\\folders"
num_files=int(input("please enter number of files want to create"))
create_folder(forlder_path, num_files)
```

```python
print('''Welcome to the simple calculator!
select an operation:
1. Addition (+)
2. Subtract (-)
3. Multiply (*)
4. divide (/) 
    ''')
def calculate(x):
    while True:
        if x=='exit':
            print("exiting the calculator. Goodbye!")
            return ""
        num1=int(input("enter first number : "))
        num2=int(input("enter second number : "))
        if x=='1':
            return num1+num2
        elif x=='2':
            return num1-num2
        elif x=='3':
            return num1*num2
        elif x=='4' and num2!=0:
            return num1/num2
        elif x=='4' and num2==0:
            print("can't devided of 0")
        else:
            print("please enter valid number")
x= input("enter your choice (1/2/3/4) or 'exit' to quit: ")
res=calculate(x)
print(res)
```
