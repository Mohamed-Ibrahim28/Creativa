Task one : Essential Operations

```python
class EssentialOperations:
    def __init__(self):
        pass

    def fact(self,n):
        while n>0:
            if n==1:
                return 1
            else :
                return n*(self.fact(n-1))


    def prime(self,n):
        if n<=1:
            return False
        elif n==2:
            return True
        if n%2==0:
            return False
        
        for i in range (3,int(n**0.5)+1,2):
            if n % i ==0:
                return False
        return True
        

    def sum_digits(self,*n):
        return sum(n)

    def power(self,base,exponent):
        return base ** exponent


class file_handling:
    def __init__(self,file_path):
        self.file_path=file_path

    def read_file(self):
        if self.file_path is None:
            return "file path is not found"
        try:
            with open(self.file_path,'r') as file:
                return file.read()
        except :
            return "file not found"
        
    def word_frequency(self):
        content =self.read_file()
        if isinstance(content,str) and content.startswith("file"):
            return content
        words= content.lower().split()
        frequency={}
        for word in words:
            word= word.strip(".,!?(){}[]\"'")
            if word in frequency:
                frequency[word]+=1
            else:
                frequency[word]=1
        return frequency



test1 = EssentialOperations()
print("the factorial is : ",test1.fact(5))
print("the number is prime ?",test1.prime(7))
print("the sum is :",test1.sum_digits(3,2,5,1))
print("the power is :",test1.power(2,3))

test2= file_handling(f"F:\\Courses\\Creativa\\session 2\\text.txt")
print(test2.read_file())
test2.word_frequency()
```
